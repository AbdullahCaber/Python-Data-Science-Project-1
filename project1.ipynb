# Questions (50p)

1. Read the csv file and create database. (10p)
2. Find the top 3 restaurants with the highest total revenue. (revenue = quantity x price) (5p)
3. Calculate the total quantity of dishes sold by category for each restaurant. (5p)
4. List the top 5 customer with highest total spending (also show total spending value) (5p)

```python
csv_path = "Data/Foodpanda Analysis Dataset.csv"
main_df = pd.read_csv(csv_path)
uni_rest_names = main_df["restaurant_name"].drop_duplicates().reset_index(drop = True)

restaurants_df = pd.DataFrame({
    "restaurant_name": uni_rest_names
    })
restaurants_df["restaurant_id"] = restaurants_df.index + 1

main_df = pd.merge(
    main_df,
    restaurants_df,
    on = "restaurant_name",
    how = "left"
)

conn = sqlite3.connect("foodpanda.db")

customer_cols = ["customer_id", "gender", "age", "city", "signup_date", "order_frequency", "last_order_date", "loyalty_points", "churned"]
customers_df = main_df[customer_cols].drop_duplicates(subset = ["customer_id"])

restaurants_df = restaurants_df[["restaurant_id", "restaurant_name"]]

order_cols = ["order_id", "customer_id", "restaurant_id", "order_date", "dish_name",
                  "category", "quantity", "price", "payment_method", "rating",
                  "rating_date", "delivery_status"]
orders_df = main_df[order_cols]

customers_df.to_sql("Customer", conn, if_exists = "replace", index = False)
restaurants_df.to_sql("Restaurant", conn, if_exists = "replace", index = False)
orders_df.to_sql("Order", conn, if_exists = "replace", index = False);
```

```python
query1 = """
SELECT
    rest_table.restaurant_name,
    SUM(order_table.quantity * order_table.price) AS total_revenue

FROM
    "Order" AS order_table

JOIN
    Restaurant AS rest_table ON order_table.restaurant_id = rest_table.restaurant_id

GROUP BY
    rest_table.restaurant_name

ORDER BY
    total_revenue DESC
LIMIT 3;
"""
top3_revenue = pd.read_sql_query(query1, conn)

print(top3_revenue)
```

```python
query2 = """
SELECT
    rest_table.restaurant_name,
    order_table.category,
    SUM(order_table.quantity) AS total_quantity

FROM
    "Order" AS order_table

JOIN
    Restaurant AS rest_table ON order_table.restaurant_id = rest_table.restaurant_id

GROUP BY
    rest_table.restaurant_name, order_table.category

ORDER BY
    rest_table.restaurant_name, total_quantity DESC;
"""
category_sales = pd.read_sql_query(query2, conn)

print(category_sales)
```

```python
query3 = """
SELECT
    customer_id,
    SUM(quantity * price) AS total_spending

FROM
    "Order"

GROUP BY
    customer_id

ORDER BY
    total_spending DESC
LIMIT 5;
"""
top5_spenders = pd.read_sql_query(query3, conn)

print(top5_spenders)
```

```python
query4 = """
SELECT
    rest_table.restaurant_name,
    AVG(order_table.rating) AS avg_rating

FROM
    "Order" AS order_table

JOIN
    Restaurant AS rest_table ON order_table.restaurant_id = rest_table.restaurant_id

GROUP BY
    rest_table.restaurant_name

ORDER BY
    avg_rating DESC;
"""
avg_ratings = pd.read_sql_query(query4, conn)

print(avg_ratings)
```

```python
query5 = """
SELECT
    customer_id,
    restaurant_id,
    order_date,
    COUNT(*) AS repetition_count

FROM
    "Order"

GROUP BY
    customer_id, restaurant_id, order_date

HAVING
    repetition_count > 1

ORDER BY
    repetition_count DESC;
"""
duplicate_orders = pd.read_sql_query(query5, conn)

if len(duplicate_orders) == 0:
    print("There is no duplicate orders in this dataset.")
else:
    print("These are the duplicate orders:")
    print(duplicate_orders)
```

```python
query6 = """
SELECT
    delivery_status,
    COUNT(*) AS status_count

FROM
    "Order"

GROUP BY
    delivery_status;
"""
status_counts_df = pd.read_sql_query(query6, conn)
total_orders = status_counts_df["status_count"].sum()
status_counts_df["Percentage"] = (status_counts_df["status_count"] / total_orders) * 100

print(status_counts_df)
```

```python
query7 = """
SELECT
    cust_table.city,
    SUM(order_table.quantity * order_table.price) AS total_revenue

FROM
    "Order" AS order_table

JOIN
    Customer AS cust_table ON order_table.customer_id = cust_table.customer_id

GROUP BY
    cust_table.city

ORDER BY
    total_revenue DESC;

"""
city_revenue_df = pd.read_sql_query(query7, conn)
total_revenue_all_cities = city_revenue_df["total_revenue"].sum()
city_revenue_df["Percentage"] = (city_revenue_df["total_revenue"] / total_revenue_all_cities) * 100

print(city_revenue_df)

conn.close()
```
